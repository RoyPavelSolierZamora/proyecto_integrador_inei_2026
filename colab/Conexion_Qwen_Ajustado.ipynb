# Conectar Qwen ajustado al dashboard EPEN
Selecciona GPU y ejecuta la única celda. La primera vez guarda el modelo y los archivos necesarios en tu Drive; en siguientes sesiones los reutiliza. No guarda encuestas ni tokens en Drive.

```python
from google.colab import drive, files
from pathlib import Path

drive.mount('/content/drive')
carpeta = Path('/content/drive/MyDrive/EPEN_Qwen_Ajustado')
carpeta.mkdir(parents=True, exist_ok=True)
script = carpeta / 'iniciar_qwen_ajustado_colab.py'

if not script.exists():
    print('Primera vez: selecciona iniciar_qwen_ajustado_colab.py desde la carpeta colab del proyecto.')
    subidos = files.upload()
    opciones = [n for n in subidos if n.startswith('iniciar_qwen_ajustado_colab') and n.endswith('.py')]
    if len(opciones) != 1:
        raise ValueError('Selecciona exactamente iniciar_qwen_ajustado_colab.py')
    with script.open('xb') as archivo:
        archivo.write(subidos[opciones[0]])

exec(compile(script.read_text(encoding='utf-8'), str(script), 'exec'), globals())
```
